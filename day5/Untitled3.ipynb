IF ELSE PROBLEMS:

```python
income = float(input("Enter annual income: "))

if income <= 250000:
    tax = 0
elif income <= 500000:
    tax = income * 0.05
elif income <= 1000000:
    tax = income * 0.20
else:
    tax = income * 0.30

print("Income Tax =", tax)
```

```python
weight=float(input("enter your weight in (kg):"))
height=float(input("enter your height in (m):"))
BMI=weight/(height*height)
print("BMI =", round(BMI,2))
if BMI<18.5:
    print("underweight")
elif BMI <29.9:
    print("normal")
elif BMI <29.9:
    print("overweight")
else:
    print("obsese")
```
