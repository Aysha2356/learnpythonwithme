```python
x=int(input("enter a number"))
y=int(input("enter a number"))
if x>0 and y>0:
    print("first quadrant")
elif x>0 and y<0:
    print("second quadrant")
elif x<0 and y<0:
    print("third quadrant")
elif x<0 and y>0:
    print("fourth quadrant")
elif x==0 and y==0:
    print("origin")
elif x==0:
    print("lies on x axis")
else:
    print("lies on y axis")
```

```python
num=int(input("enter a number:"))
temp=num
rev=0
while temp >0:
    digit=temp%10
    rev=rev*10+digit
    temp=temp//10
if num++rev:
    print("palindrome")
else:
    print("not a palindrome")
    
```

```python
num=int(input("enter a number"))
temp=num
digits=len(str(num))
total=0
while temp > 0:
    digit=temp%10
    total=total+digit**digits
    temp=temp//10
if total==num:
    print("armstrong")
else:
    print("not a armstrong")
```

```python
password = input("Enter password: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True
    else:
        has_special = True

if len(password) >= 8 and has_upper and has_lower and has_digit and has_special:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
num1=int(input("enter a number"))
num2=int(input("enter a number"))
choice=input("enter a operator(+,-,*,%):")
match choice:
    case "+":
        print("the result:",num1+num2)
    case "-":
        print("the result:",num1-num2)
    case "*":
        print("the result:",num1*num2)
    case "%":
        print("the result:",num1%num2)
```
